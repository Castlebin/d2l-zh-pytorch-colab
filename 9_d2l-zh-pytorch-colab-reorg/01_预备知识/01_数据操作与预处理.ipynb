## 第一部分: 张量数据操作

### 1.1 张量简介

张量(tensor)是深度学习框架的核心数据结构。它是一个$n$维数组,可以在GPU上加速计算,并支持自动微分。

- 1维张量 = 向量(vector)
- 2维张量 = 矩阵(matrix)  
- 3维及以上张量 = 高维张量

首先导入PyTorch库:

```python
import torch
```

### 1.2 创建张量

**创建行向量**

```python
# 使用 arange 创建一个包含0到11的行向量
x = torch.arange(12)
x
```

**查看张量的形状和元素总数**

```python
# 查看张量形状
print("形状:", x.shape)

# 查看元素总数
print("元素总数:", x.numel())
```

**改变张量形状**

```python
# 将形状从(12,)改为(3, 4)
X = x.reshape(3, 4)
print(X)

# 使用-1自动计算维度
X_auto = x.reshape(-1, 4)  # 自动计算为3行
print("\n自动计算维度:")
print(X_auto)
```

**创建特殊张量**

```python
# 全0张量
zeros = torch.zeros((2, 3, 4))
print("全0张量:")
print(zeros)

# 全1张量
ones = torch.ones((2, 3, 4))
print("\n全1张量:")
print(ones)

# 随机张量(标准正态分布)
random_tensor = torch.randn((3, 4))
print("\n随机张量:")
print(random_tensor)

# 从Python列表创建张量
tensor_from_list = torch.tensor([[2, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
print("\n从列表创建:")
print(tensor_from_list)
```

### 1.3 张量运算

**按元素运算**

```python
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])

print("加法:", x + y)
print("减法:", x - y)
print("乘法:", x * y)
print("除法:", x / y)
print("幂运算:", x ** y)
print("指数运算:", torch.exp(x))
```

**矩阵运算**

```python
# 张量拼接
X = torch.arange(12, dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])

print("原张量X:")
print(X)
print("\n原张量Y:")
print(Y)

# 按行拼接(沿着第0维)
print("\n按行拼接:")
print(torch.cat((X, Y), dim=0))

# 按列拼接(沿着第1维)
print("\n按列拼接:")
print(torch.cat((X, Y), dim=1))
```

**逻辑运算**

```python
# 通过逻辑运算构建二元张量
print("X == Y:")
print(X == Y)
```

**求和与统计**

```python
# 所有元素求和
print("所有元素的和:", X.sum())

# 其他统计函数
print("平均值:", X.mean())
print("最大值:", X.max())
print("最小值:", X.min())
```

### 1.4 索引和切片

```python
# 访问单个元素
print("最后一行:", X[-1])
print("第2行第3列:", X[1, 2])

# 切片访问
print("\n第2-3行,第1-3列:")
print(X[1:3, 0:3])

# 修改元素
X[1, 2] = 9
print("\n修改后的X:")
print(X)

# 修改多个元素
X[0:2, :] = 12
print("\n批量修改后的X:")
print(X)
```

### 1.5 内存管理与运算

```python
# 查看内存地址
before = id(Y)
Y = Y + X
print("内存地址改变:", id(Y) != before)

# 原地操作(不分配新内存)
Z = torch.zeros_like(Y)
print("原地操作前ID:", id(Z))
Z[:] = X + Y
print("原地操作后ID:", id(Z))

# 使用 += 等操作也是原地的
before = id(X)
X += Y
print("\n使用+=后ID不变:", id(X) == before)
```

### 1.6 与其他Python对象的转换

```python
# 转换为NumPy数组
A = X.numpy()
print("NumPy数组:")
print(type(A))
print(A)

# 从NumPy数组转换为张量
B = torch.tensor(A)
print("\n转回张量:")
print(type(B))

# 标量张量转换为Python数值
a = torch.tensor([3.5])
print("\n标量张量:", a, a.item(), type(a.item()))
```

### 1.7 广播机制

```python
# 不同形状的张量可以通过广播机制进行运算
a = torch.arange(3).reshape((3, 1))
b = torch.arange(2).reshape((1, 2))

print("a的形状:", a.shape)
print(a)
print("\nb的形状:", b.shape)
print(b)

print("\na + b (广播机制):")
print(a + b)
```

## 第二部分: 数据预处理

### 2.1 使用Pandas处理表格数据

在实际项目中,我们经常需要处理CSV、Excel等格式的原始数据。Pandas是Python中最常用的数据分析工具。

```python
import os
import pandas as pd
```

### 2.2 创建和读取数据集

```python
# 创建数据目录
os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')

# 写入CSV文件
with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n')  # 列名
    f.write('NA,Pave,127500\n')  # 每行表示一个数据样本
    f.write('2,NA,106000\n')
    f.write('4,NA,178100\n')
    f.write('NA,NA,140000\n')

print("CSV文件已创建")
```

```python
# 读取CSV文件
data = pd.read_csv(data_file)
print("原始数据:")
print(data)
```

### 2.3 处理缺失值

真实数据经常包含缺失值(NaN)。常见的处理方法:
- **插值法**: 用其他值(如均值)填充
- **删除法**: 删除含缺失值的行或列

```python
# 将数据分为输入和输出
inputs, outputs = data.iloc[:, 0:2], data.iloc[:, 2]

# 用均值填充数值列的缺失值

# inputs = inputs.fillna(inputs.mean()) # 报错了，数据类型不匹配，因为有非数值
# 修改为
# 用均值填充数值列的缺失值，并使用numeric_only=True确保只对数值列进行操作
# 非数值列的缺失值（如'Alley'列）将保留，后续由pd.get_dummies处理
inputs = inputs.fillna(inputs.mean(numeric_only=True))


print("填充后的inputs:")
print(inputs)
```

### 2.4 处理类别数据

对于类别特征,我们使用独热编码(one-hot encoding)将其转换为数值。

```python
# 对类别列进行独热编码
inputs = pd.get_dummies(inputs, dummy_na=True)
print("独热编码后的inputs:")
print(inputs)
```

### 2.5 转换为张量格式

```python
# 将处理好的数据转换为张量
X = torch.tensor(inputs.to_numpy(dtype=float))
y = torch.tensor(outputs.to_numpy(dtype=float))

print("特征张量X:")
print(X)
print("\n标签张量y:")
print(y)
```

```python
# @title 练习 1. 创建一个5×5的随机矩阵,计算它的转置和行列式

import torch

X = torch.rand(5, 5)
print("矩阵 X=", X)

# 矩阵的转置
X_t = X.T
print("矩阵 X 的转置 X.T=", X_t)

# 矩阵的行列式
X_det = X.det()
print("矩阵 X 的行列式 X.det()=", X_det)
```

```python
# @title 练习 2. 实现一个函数,将摄氏温度数组转换为华氏温度

## 方法 1：基础 Python（使用列表推导式）
def celsius_to_fahrenheit(celsius_list):
    """
    将摄氏度列表转换为华氏度列表
    公式: F = C * 9/5 + 32
    """
    return [round(c * 9/5 + 32, 2) for c in celsius_list]

# 测试
c_temps = [0, 25, 37, 100]
f_temps = celsius_to_fahrenheit(c_temps)
print(f"华氏温度: {f_temps}")
# 输出: [32.0, 77.0, 98.6, 212.0]



## 方法 2：使用 NumPy（适合大规模数据）
# 如果你处理的是大型数组，NumPy 的向量化操作速度会快得多，且代码最简洁。
import numpy as np

def convert_with_numpy(celsius_array):
    # 直接对整个数组进行数学运算
    c_arr = np.array(celsius_array)
    return c_arr * 9/5 + 32

# 测试
c_temps = [0, 25, 37, 100]
print(f"NumPy 转换结果: {convert_with_numpy(c_temps)}")

```

```python
# @title 练习 3. 创建包含更多行和列的数据集：
## 删除缺失值最多的列、将预处理后的数据集转换为张量格式

# 1. 创建大型数据集
# 使用 pandas 和 numpy 可以快速生成具有大量行和列的模拟数据，并随机插入缺失值。
import pandas as pd
import numpy as np
import torch

# 创建一个 1000 行 10 列的大型数据集
rows, cols = 1000, 10
data = np.random.randn(rows, cols)
df = pd.DataFrame(data, columns=[f'Col_{i}' for i in range(cols)])

# 随机插入一些缺失值 (NaN)
# 让 Col_3 成为缺失值最多的列
df.iloc[:500, 3] = np.nan
df.iloc[:100, 5] = np.nan


# 2. 删除缺失值最多的列
# 通过 isnull().sum() 计算每列的缺失值数量，找到最大值对应的索引并删除。
# 找到缺失值数量最多的列名
col_with_most_nan = df.isnull().sum().idxmax()
print(f"缺失值最多的列是: {col_with_most_nan}, 数量: {df[col_with_most_nan].isnull().sum()}")

# 删除该列
df_cleaned = df.drop(columns=[col_with_most_nan])


# 3. 将数据集转换为张量 (Tensor)
# 在转换之前，必须确保数据集中不再包含缺失值（因为 PyTorch 张量不支持在运算中包含 NaN，通常会导致梯度消失或报错）。一般采用均值填充或其他预处理。
# 3.1. 预处理：填充剩余的缺失值（例如用均值填充）
df_preprocessed = df_cleaned.fillna(df_cleaned.mean())

# 3.2. 转换为张量
# 使用 .values 获取 numpy 数组，再转为 tensor
# 推荐指定 dtype 为 float32，这是深度学习最常用的精度
data_tensor = torch.tensor(df_preprocessed.values, dtype=torch.float32)

print("张量形状:", data_tensor.shape)
print("前两行数据:\n", data_tensor[:2])


'''
关键步骤总结
1. isnull().sum(): 用于统计每列的缺失值分布。
2. drop(): 根据列名删除特定维度的数据。
3. fillna(): 在转换前必须处理掉所有 NaN，否则张量运算会出错。
4. torch.tensor(): 将处理好的 numpy 或 DataFrame 数组直接转换为 PyTorch 张量。
'''
```

```python
# @title 练习 4. 尝试使用广播机制,对一个矩阵的每一行减去该行的均值
import torch

X = torch.arange(12, dtype=float).reshape(3, 4)
print("矩阵 X: ", X)

# 2. 计算每一行的均值
# dim=1 表示跨列计算（即对每一行求均值）
# keepdim=True 是关键：它将结果形状从 [3] 变为 [3, 1]
row_means = X.mean(dim=1, keepdim=True)
print("\n每行的均值 (形状 [3, 1]):\n", row_means)

# 3. 利用广播机制进行减法
# X 的形状是 [3, 4]，row_means 的形状是 [3, 1]
# 广播机制会自动将 [3, 1] 扩展为 [3, 4] 以匹配 X
X_centered = X - row_means

print("\n减去行均值后的结果:\n", X_centered)

# 验证：现在每一行的和（或均值）应该接近 0
print("\n验证（每行求和）:\n", X_centered.sum(dim=1))
```
