## 第一部分: 张量数据操作

### 1.1 张量简介

张量(tensor)是深度学习框架的核心数据结构。它是一个$n$维数组,可以在GPU上加速计算,并支持自动微分。

- 1维张量 = 向量(vector)
- 2维张量 = 矩阵(matrix)  
- 3维及以上张量 = 高维张量

首先导入PyTorch库:

```python
import torch
```

### 1.2 创建张量

**创建行向量**

```python
# 使用 arange 创建一个包含0到11的行向量
x = torch.arange(12)
x
```

**查看张量的形状和元素总数**

```python
# 查看张量形状
print("形状:", x.shape)

# 查看元素总数
print("元素总数:", x.numel())
```

**改变张量形状**

```python
# 将形状从(12,)改为(3, 4)
X = x.reshape(3, 4)
print(X)

# 使用-1自动计算维度
X_auto = x.reshape(-1, 4)  # 自动计算为3行
print("\n自动计算维度:")
print(X_auto)
```

```python
# 自动计算列为 4 列
X_auto_2 = x.reshape(3, -1)
print(X_auto_2)
```

```python
# 自动计算一个中间的维度
X_auto_3 = x.reshape(2, -1, 3)
print(X_auto_3)
```

**创建特殊张量**

```python
# 全0张量
zeros = torch.zeros((2, 3, 4))
print("全0张量:")
print(zeros)

# 全1张量
ones = torch.ones((2, 3, 4))
print("\n全1张量:")
print(ones)

# 随机张量(标准正态分布) torch.randn()
random_tensor = torch.randn((3, 4))
print("\n随机张量:")
print(random_tensor)

# 随机张量 torch.rand()
rand_2 = torch.rand(3, 4)
print("\n随机张量 2 :", rand_2)

# 从Python列表创建张量
tensor_from_list = torch.tensor([[2, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
print("\n从列表创建:")
print(tensor_from_list)
```

### 1.3 张量运算

**按元素运算**

```python
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])

print("加法:", x + y)
print("减法:", x - y)
print("乘法:", x * y)
print("除法:", x / y)
print("幂运算:", x ** y)
print("指数运算:", torch.exp(x))
```

**矩阵运算**

```python
# 张量拼接
X = torch.arange(12, dtype=torch.float32).reshape((3,4))
Y = torch.tensor([
    [2.0, 1, 4, 3],
    [1, 2, 3, 4],
    [4, 3, 2, 1]
])

print("原张量X:")
print(X)
print("\n原张量Y:")
print(Y)

# 按行拼接(沿着第0维)
print("\n按行拼接:")
print(torch.cat((X, Y), dim=0))

# 按列拼接(沿着第1维)
print("\n按列拼接:")
print(torch.cat((X, Y), dim=1))
```

**逻辑运算**

```python
# 通过逻辑运算构建二元张量
print("X == Y:")
print(X == Y)
```

**求和与统计**

```python
# 所有元素求和
print("所有元素的和:", X.sum())

# 其他统计函数
print("平均值:", X.mean())
print("最大值:", X.max())
print("最小值:", X.min())
```

### 1.4 索引和切片

```python
# 访问单个元素
print("最后一行:", X[-1])
print("第2行第3列:", X[1, 2])

# 切片访问
print("\n第2-3行,第1-3列:")
print(X[1:3, 0:3])

# 修改元素
X[1, 2] = 9
print("\n修改后的X:")
print(X)

# 修改多个元素
X[0:2, :] = 12
print("\n批量修改后的X:")
print(X)
```

### 1.5 内存管理与运算

```python
# 查看内存地址
before = id(Y)
Y = Y + X
print("内存地址改变:", id(Y) != before)

# 原地操作(不分配新内存)
Z = torch.zeros_like(Y)
print("原地操作前ID:", id(Z))
Z[:] = X + Y
print("原地操作后ID:", id(Z))

# 使用 += 等操作也是原地的
before = id(X)
X += Y
print("\n使用+=后ID不变:", id(X) == before)
```

### 1.6 与其他Python对象的转换

```python
# 转换为NumPy数组
A = X.numpy()
print("NumPy数组:")
print(type(A))
print(A)

# 从NumPy数组转换为张量
B = torch.tensor(A)
print("\n转回张量:")
print(type(B))

# 标量张量转换为Python数值
a = torch.tensor([3.5])
print("\n标量张量:", a, a.item(), type(a.item()))
```

### 1.7 广播机制

```python
# 不同形状的张量可以通过广播机制进行运算
a = torch.arange(3).reshape((3, 1))
b = torch.arange(2).reshape((1, 2))

print("a的形状:", a.shape)
print(a)
print("\nb的形状:", b.shape)
print(b)

print("\na + b (广播机制):")
print(a + b)

# 可以看到，它是将两个矩阵维度扩展后相加
```

## 第二部分: 数据预处理

### 2.1 使用Pandas处理表格数据

在实际项目中,我们经常需要处理CSV、Excel等格式的原始数据。Pandas是Python中最常用的数据分析工具。

```python
import os
import pandas as pd
```

### 2.2 创建和读取数据集

```python
# 创建数据目录
os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')

# 写入CSV文件
with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n')  # 列名
    f.write('NA,Pave,127500\n')  # 每行表示一个数据样本
    f.write('2,NA,106000\n')
    f.write('4,NA,178100\n')
    f.write('NA,NA,140000\n')

print("CSV文件已创建")
```

```python
# 读取CSV文件
data = pd.read_csv(data_file)
print("原始数据:")
print(data)
```

### 2.3 处理缺失值

真实数据经常包含缺失值(NaN)。常见的处理方法:
- **插值法**: 用其他值(如均值)填充
- **删除法**: 删除含缺失值的行或列

```python
# 将数据分为输入和输出
inputs, outputs = data.iloc[:, 0:2], data.iloc[:, 2]

# 用均值填充数值列的缺失值
# inputs = inputs.fillna(inputs.mean())  # 报错，数据类型问题！fixed

# 用均值填充数值列的缺失值，并使用numeric_only=True确保只对数值列进行操作
# 非数值列的缺失值（如'Alley'列）将保留，后续由pd.get_dummies处理
inputs = inputs.fillna(inputs.mean(numeric_only=True))

print("填充后的inputs:")
print(inputs)
```

### 2.4 处理类别数据

对于类别特征,我们使用独热编码(one-hot encoding)将其转换为数值。

```python
# 对类别列进行独热编码
inputs = pd.get_dummies(inputs, dummy_na=True)
print("独热编码后的inputs:")
print(inputs)
```

### 2.5 转换为张量格式

```python
# 将处理好的数据转换为张量
X = torch.tensor(inputs.to_numpy(dtype=float))
y = torch.tensor(outputs.to_numpy(dtype=float))

print("特征张量X:")
print(X)
print("\n标签张量y:")
print(y)
```
