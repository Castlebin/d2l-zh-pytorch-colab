# 云端硬件与 GPU 选择指南

本笔记结合原 `selecting-servers-gpus.ipynb` 与 `aws.ipynb`,整理深度学习中常见的硬件、云平台选择策略,帮助你在成本与性能之间做出合理取舍。

## 1. 选择依据

| 维度 | 关键问题 | 建议 |
|------|----------|------|
| 任务类型 | 视觉 / NLP / 生成模型 | 视觉偏向高显存, NLP 偏向高显存 + 高带宽 |
| 数据规模 | 本地能否存储? | TB 级建议云端,GB 级可本地 |
| 预算 | 一次性采购 or 按需租用 | 学习/短期项目推荐租用 |
| 升级周期 | 是否需随时切换硬件 | 云端灵活,本地需要提前规划 |

## 2. GPU 型号速览

| GPU | 显存 | FP16 算力 | 适用场景 |
|-----|------|-----------|-----------|
| RTX 3060 | 12 GB | 24 TFLOPS | 入门学习、CV 任务 |
| RTX 4090 | 24 GB | 82 TFLOPS | 大模型微调、生成任务 |
| A10G | 24 GB | 31 TFLOPS | 云端按需,适合推理 |
| A100 40G | 40 GB | 156 TFLOPS | 大规模训练、混合精度 |
| A100 80G | 80 GB | 312 TFLOPS | 超大模型、分布式训练 |

> ⚠️ 选择 GPU 时优先考虑显存;算力不足可以通过更长训练时间弥补,显存不足会导致无法启动。

## 3. 云服务厂商对比

| 平台 | 优势 | 注意事项 |
|------|------|----------|
| AWS | 全球节点多,生态完善 | 价格相对较高,首次配置略复杂 |
| Google Cloud | 与 Colab 深度整合 | 某些 GPU 需配额申请 |
| Azure | 与微软工具链整合 | 文档偏企业级 |
| Paperspace | 价格透明,适合个人 | 网络带宽需额外购买 |
| Lambda Cloud | 面向深度学习优化 | 支付方式主要为信用卡 |

## 4. 建议的配置模板

### 4.1 学习 / 小型实验

- GPU: RTX 3060 / T4 / L4
- vCPU: 4~8 核
- 内存: 16~32 GB
- 存储: 100 GB SSD
- 典型费用: $0.3~0.6 / 小时

### 4.2 中等规模训练

- GPU: RTX 3090 / A10G
- vCPU: 16 核以上
- 内存: 64 GB
- 存储: 500 GB SSD + 对象存储
- 典型费用: $1~3 / 小时

### 4.3 大模型/多 GPU

- GPU: A100 40G/80G, H100
- 方式: 多机多卡或单机 4 卡
- 典型费用: $10+/小时,需申请配额

## 5. 成本优化策略

1. **抢占式实例 / Spot Instance**: 价格可低 50%+,适合可中断任务。
2. **预留实例**: 适合长期项目,按年预付获得折扣。
3. **数据分层存储**: 热数据放 NVMe,冷数据放对象存储(S3、OSS)。
4. **自动化关机**: 训练结束自动释放实例,防止忘记关机产生账单。
5. **日志与模型分离**: 大文件上传至对象存储,减少本地磁盘开销。

## 6. 网络与安全

- 启用 SSH 密钥登录,禁用密码。
- 使用 `tmux` / `screen` 保障训练不中断。
- 对外暴露 Notebook 时务必加密码或通过 `ssh -L` 做端口转发。
- 经常更新系统补丁,保持显卡驱动与 CUDA 匹配。

## 7. 购买本地 GPU 的场景

- 需要离线环境(实验室、企业内网)。
- 硬件可长期使用(3 年以上)。
- 对数据隐私要求严格,无法上传云端。

建议配置: i9/Ryzen 9 + 64 GB 内存 + 2 TB NVMe + 显卡(根据预算选择)。
