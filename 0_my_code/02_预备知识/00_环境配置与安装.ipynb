# 环境配置与安装

本笔记整理了在本地或 Google Colab 上准备深度学习环境的完整流程。所有步骤都避免依赖 `d2l` 包,仅使用 PyTorch 及常见数据科学库,从而保证与 Colab 兼容。

## 1. Python 与虚拟环境

- 推荐 Python 版本: **3.8~3.11**。如使用 Conda,可执行:
  ```bash
  conda create -n d2l python=3.10
  conda activate d2l
  ```
- 纯 pip 用户可以使用 `python -m venv venv && source venv/bin/activate` 创建隔离环境。
- 升级包管理工具:
  ```bash
  python -m pip install --upgrade pip setuptools wheel
  ```

## 2. 安装核心依赖

| 库 | 说明 | 安装命令 |
|----|------|----------|
| PyTorch | 深度学习框架 | 见下文,需选择 CUDA 版本 |
| torchvision | 视觉数据集与模型 | `pip install torchvision` |
| torchaudio | 音频处理(可选) | `pip install torchaudio` |
| matplotlib | 可视化 | `pip install matplotlib` |
| pandas | 表格数据处理 | `pip install pandas` |
| scikit-learn | 评估与数据预处理 | `pip install scikit-learn` |
| jupyterlab / notebook | 交互式实验 | `pip install notebook` |

### 2.1 安装 PyTorch

- 访问 [PyTorch 官网](https://pytorch.org/get-started/locally/) 选择对应操作系统、包管理器和 CUDA 版本,复制命令执行即可。
- 纯 CPU 环境可以直接安装:
  ```bash
  pip install torch --index-url https://download.pytorch.org/whl/cpu
  ```
- Colab 自带 GPU 版本,无需额外安装,执行 `import torch` 即可使用。

## 3. 验证安装

执行下方代码单元,确认 PyTorch 与 CUDA 的可用性。



````python
import torch
import platform

print(f'Python 版本: {platform.python_version()}')
print(f'PyTorch 版本: {torch.__version__}')
print('CUDA 是否可用:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU 数量:', torch.cuda.device_count())
    print('当前 GPU:', torch.cuda.get_device_name(0))

````

## 5. 常见问题排查

1. **ImportError: libtorch_cuda.so not found**
   - 原因: CUDA 版本与驱动不兼容。
   - 解决: 卸载后使用官网命令重新安装,或改用 CPU 版本。

2. **No module named 'torch'**
   - 原因: 当前解释器未安装。
   - 解决: 激活正确的虚拟环境后重新安装。

3. **显存不足**
   - 解决: 减小 `batch_size`, 使用 `torch.cuda.empty_cache()`, 或启用混合精度 (`torch.cuda.amp`).
