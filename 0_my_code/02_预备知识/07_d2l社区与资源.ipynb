# d2l 社区与学习资源

该笔记基于原 `d2l.ipynb` 与 `contributing.ipynb`,介绍《动手学深度学习(Dive into Deep Learning)》项目的背景、生态资源以及参与贡献的方式。

## 1. 项目简介

- **官方网站**: [https://d2l.ai](https://d2l.ai)
- **主要作者**: Aston Zhang、Mu Li、Zachary C. Lipton 等
- **核心目标**: 通过交互式 Jupyter Notebook 将数学、代码与实践结合,降低深度学习入门门槛。

《动手学深度学习》以 PyTorch、TensorFlow、MXNet 等多种实现提供教学示例。本仓库为 PyTorch+中文重构版本,已经完成独立化,无需 `d2l` 包。

## 3. 学习路线建议

1. 预备知识 → 线性模型 → 优化方法 → 卷积/循环/注意力
2. 每章至少完成一次“从零实现”和“PyTorch 实现”对比
3. 配合 Kaggle 实战章节,将模型应用到真实数据集
4. 完成练习题并在社区分享解题思路

## 4. 如何向社区贡献

1. **Fork 仓库** → 创建功能分支
2. 完成修改后运行 `pytest` 或相应脚本确保通过
3. 提交 Pull Request,并在描述中附上:
   - 修改动机
   - 测试方式
   - 截图或日志(如涉及可视化)
4. 与维护者讨论,根据反馈更新 PR

### 4.1 适合新手的贡献方式

- 修正错别字、翻译或格式
- 优化 Notebook 说明文字、补充练习答案
- 更新依赖版本并测试兼容性
- 将实验结果记录在 README 或教程中

### 4.2 高阶贡献方向

- 添加新章节或新案例(如扩展 Transformer 实战)
- 增加动手项目或 Kaggle 竞赛的完整解法
- 重构测试框架、提高训练脚本的可复现性

## 6. 常见问题

1. **如何同步上游更新?**
   ```bash
   git remote add upstream https://github.com/d2l-ai/d2l-zh.git
   git fetch upstream
   git merge upstream/master
   ```
2. **提交 PR 需要跑全部 Notebook 吗?**
   - 建议至少运行受影响的 Notebook
   - 若涉及依赖变更,应在本地与 Colab 各跑一遍
3. **是否可以引用本仓库内容?**
   - 可以,请注明来源并遵循 MIT 许可
