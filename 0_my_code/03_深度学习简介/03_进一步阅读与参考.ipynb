## 6. 论文阅读建议

1. 先读摘要和结论,判断是否与当前任务相关。
2. 关注模型结构图与伪代码,了解核心贡献。
3. 将论文中的符号与本仓库 Notebook 对照,加深理解。
4. 尝试复现实验,或将论文中的想法融入已有项目。

## 7. 引用本项目

如果你基于本仓库发表文章或项目,可使用如下引用格式:
