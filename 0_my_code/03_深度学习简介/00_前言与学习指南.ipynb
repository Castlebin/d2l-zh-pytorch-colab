# 前言与学习指南

本文改写自原仓库 `chapter_preface/index.ipynb`,介绍课程背景、阅读方式及如何高效使用重构后的中文代码库。

## 1. 为什么选择《动手学深度学习》

- **理论 + 实践**: 每章都从数学推导起步,立即给出可运行的 PyTorch 实现。
- **可交互**: Notebook 允许逐行修改、可视化,适合实验与教学。
- **开放社区**: 大量读者在 GitHub 与论坛分享改进方案与习题答案。

## 2. 适用人群

| 背景 | 建议 |
|------|------|
| 编程入门 | 先完成 Python 基础、Numpy、Pandas 课程 |
| ML 初学者 | 按章节顺序学习,不跳过练习 |
| 工程实践者 | 快速浏览理论,重点关注“实战” Notebook |
| 研究人员 | 阅读高级章节(注意力、预训练、应用)并尝试扩展 |

## 3. 如何使用本仓库

1. **逐章阅读**: 中文目录已按难度排序,横向递进。
2. **运行 Notebook**: 推荐在本地/Colab 复制仓库并运行 `pip install -r requirements.txt`。
3. **动手实验**: 对每个模型尝试修改参数、替换数据集或添加可视化。
4. **写笔记**: 建议记录每章关键概念、公式推导与实验结果。
5. **完成练习**: Notebook 末尾的练习可帮助检验理解。

## 5. 学习建议

- **先跑通官方代码** 再尝试封装和优化。
- **随手可视化**: 利用 `matplotlib`/`seaborn` 观察 loss、accuracy 曲线。
- **多读论文**: 每章附带的参考文献提供原始论文,可扩展理解。
- **复盘总结**: 每次学习结束写下“学到了什么”“还有哪些疑问”。
