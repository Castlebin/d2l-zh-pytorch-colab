## 8. 在 Notebook 中快速引用

以下小工具可在 Notebook 顶部粘贴,便于快速展示符号。



```python
SYMBOLS = {
    'relu': 'ReLU 激活函数: relu(x) = max(0, x)',
    'softmax': 'Softmax: exp(z_i) / sum_j exp(z_j)',
    'cross_entropy': '交叉熵: -sum_i y_i log(\hat{y}_i)',
}

def show_symbol(name):
    desc = SYMBOLS.get(name)
    if desc:
        print(f'{name}: {desc}')
    else:
        print('未找到该符号,请补充到 SYMBOLS 字典中。')

show_symbol('relu')

```
