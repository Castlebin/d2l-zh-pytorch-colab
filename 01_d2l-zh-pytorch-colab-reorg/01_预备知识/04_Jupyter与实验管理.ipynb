# Jupyter 与实验管理

本笔记总结如何高效使用 Jupyter Notebook/JupyterLab 进行深度学习实验,涵盖常用快捷键、调试方法以及与 Git 配合的技巧,替代原 `jupyter.ipynb`。

## 1. Jupyter 快速上手

### 1.1 安装与启动

```bash
pip install notebook jupyterlab
jupyter lab  # 或者 jupyter notebook
```

- `jupyter lab` 提供标签页、多文档界面,更适合大型项目。
- 在 Colab 中直接新建 Notebook 即可,无需本地安装。

### 1.2 常用快捷键

| 模式 | 快捷键 | 功能 |
|------|--------|------|
| 命令模式 | `A` / `B` | 在上方 / 下方插入单元 |
| 命令模式 | `D` `D` | 删除单元 |
| 命令模式 | `M` / `Y` | 切换为 Markdown / 代码单元 |
| 编辑模式 | `Ctrl`+`Enter` | 运行当前单元 |
| 编辑模式 | `Shift`+`Enter` | 运行并跳到下一单元 |
| 任意 | `Esc` | 进入命令模式 |

熟练掌握快捷键可以显著提高笔记运行效率。

## 2. 高效组织 Notebook

1. **标题结构**: 使用 `#`、`##`、`###` 划分章节,便于生成目录。
2. **最少的全局状态**: 将所有依赖定义在顶端,避免跨单元“隐式变量”。
3. **可重复运行**: 确保从“重启内核并运行全部”可以顺利执行。
4. **数据路径变量化**: 使用 `DATA_DIR = Path('data')` 管理路径,兼容本地与 Colab。
5. **自动化实验记录**: 建议在末尾保存关键超参数与指标,可配合 `json` 或 `yaml` 导出。

## 3. 在 Notebook 中调试

- **断点调试**: 利用 `%debug` 魔法命令进入交互式调试,或在 Python 3.7+ 中使用 `breakpoint()`。
- **性能分析**: 使用 `%%time` / `%%timeit` 测量运行时间,或 `torch.profiler` 进行更细致分析。
- **日志输出**: 借助 `logging` 库记录训练信息,避免刷屏。

```python
import logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)
logger.info('开始训练')
```

## 4. 与 Git 协作

- 使用 `nbstripout` 或 `jupyter nbconvert` 清理输出,减少仓库体积。
- 版本控制大文件建议存储在 `outputs/` 并写入 `.gitignore`。
- 需要审阅差异时,可以运行:

```bash
jupyter nbconvert --to notebook --ClearOutputPreprocessor.enabled=True notebook.ipynb
nbdiff notebook_old.ipynb notebook_new.ipynb
```

## 5. Notebook 转脚本

当实验稳定后,可将 Notebook 转换为 Python 脚本,方便批量训练或部署。

```bash
jupyter nbconvert notebook.ipynb --to python --output train.py
```

转换后的脚本可以整合进 CI/CD 流程,保证训练可复现。

## 6. 推荐扩展

| 扩展 | 功能 |
|------|------|
| `nbextensions` | 提供目录、变量检查器等实用功能 |
| `jupyterlab_code_formatter` | 一键格式化代码 |
| `ipywidgets` | 构建交互式控件,用于调参或可视化 |

安装示例:

```bash
pip install jupyterlab_code_formatter black isort
jupyter labextension install @ryantam626/jupyterlab_code_formatter
```

## 7. Notebook 到 Colab

- 将仓库推送到 GitHub 后,使用 `https://colab.research.google.com/github/<user>/<repo>/blob/main/path.ipynb` 即可在 Colab 打开。
- 建议使用 `%pip install -r requirements.txt` 在首个单元安装依赖。
